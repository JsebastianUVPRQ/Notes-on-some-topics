## 1. Ejercicio 11.2.10: Valores Complejos donde las Funciones tienen Derivada

Para analizar funciones con exponentes fraccionarios o funciones trigonométricas/hiperbólicas inversas en el plano complejo, recordamos que son **funciones multivaluadas** definidas mediante el logaritmo complejo. Para ser analíticas (univaluadas y derivables), requieren seleccionar un **corte de rama**.

### (a) $f(z) = z^{3/2}$
* **Expresión logarítmica:** $f(z) = e^{\frac{3}{2} \log z}$
* **Derivada formal:** $f'(z) = \frac{3}{2} z^{1/2}$
* **Dominio de derivabilidad:** $z = 0$ es un punto de ramificación. Asumiendo la rama principal (corte en el semieje real negativo), la función es derivable en:
  $$\mathbb{C} \setminus (-\infty, 0]$$

### (b) $f(z) = z^{-3/2}$
* **Expresión logarítmica:** $f(z) = e^{-\frac{3}{2} \log z}$
* **Derivada formal:** $f'(z) = -\frac{3}{2} z^{-5/2} = -\frac{3}{2 z^{5/2}}$
* **Dominio de derivabilidad:** $z = 0$ es punto de ramificación y singularidad. Bajo la rama principal, es derivable en:
  $$\mathbb{C} \setminus (-\infty, 0]$$

### (c) $f(z) = \tan^{-1}(z)$
* **Expresión logarítmica:** $f(z) = \frac{1}{2i} \log\left(\frac{1+iz}{1-iz}\right)$
* **Derivada formal:** $f'(z) = \frac{1}{1+z^2}$
* **Puntos de ramificación:** $z = \pm i$
* **Dominio de derivabilidad (rama principal):** Todo $\mathbb{C}$ excepto los cortes sobre el eje imaginario:
  $$\mathbb{C} \setminus \left( [i, i\infty) \cup (-i\infty, -i] \right)$$

### (d) $f(z) = \tanh^{-1}(z)$
* **Expresión logarítmica:** $f(z) = \frac{1}{2} \log\left(\frac{1+z}{1-z}\right)$
* **Derivada formal:** $f'(z) = \frac{1}{1-z^2}$
* **Puntos de ramificación:** $z = \pm 1$
* **Dominio de derivabilidad (rama principal):** Todo $\mathbb{C}$ excepto los cortes sobre el eje real:
  $$\mathbb{C} \setminus \left( (-\infty, -1] \cup [1, \infty) \right)$$

```python
# Verificación simbólica con SymPy de las derivadas
import sympy as sp

z = sp.Symbol('z', complex=True)

f_a = z**(3/2)
f_b = z**(-3/2)
f_c = sp.atan(z)
f_d = sp.atanh(z)

print("(a) d/dz(z^(3/2))   =", sp.diff(f_a, z))
print("(b) d/dz(z^(-3/2))  =", sp.diff(f_b, z))
print("(c) d/dz(atan(z))   =", sp.diff(f_c, z))
print("(d) d/dz(atanh(z))  =", sp.diff(f_d, z))
```

## 2. Ejercicio 11.2.11: Potencial Complejo en Dinámica de Fluidos

Un flujo bidimensional irrotacional se describe con un potencial complejo $f(z) = u(x,y) + iv(x,y)$, donde $u$ es el potencial de velocidad y $v$ es la función de corriente. La velocidad es $\mathbf{V} = \nabla u = \left(\frac{\partial u}{\partial x}, \frac{\partial u}{\partial y}\right)$.

### (a) Demostrar que $\frac{df}{dz} = V_x - iV_y$

**Demostración:**
1. Las componentes de velocidad son $V_x = \frac{\partial u}{\partial x}$ y $V_y = \frac{\partial u}{\partial y}$.
2. Como $f(z)$ es analítica, derivamos a lo largo del eje $x$:
   $$\frac{df}{dz} = \frac{\partial u}{\partial x} + i \frac{\partial v}{\partial x}$$
3. Aplicamos las ecuaciones de Cauchy-Riemann ($\frac{\partial v}{\partial x} = -\frac{\partial u}{\partial y}$):
   $$\frac{df}{dz} = \frac{\partial u}{\partial x} - i \frac{\partial u}{\partial y}$$
4. Sustituyendo $V_x$ y $V_y$:
   $$\frac{df}{dz} = V_x - iV_y \quad \blacksquare$$

---

### (b) Demostrar que $\nabla \cdot \mathbf{V} = 0$ (Incompresibilidad)

**Demostración:**
1. Divergencia: $\nabla \cdot \mathbf{V} = \frac{\partial V_x}{\partial x} + \frac{\partial V_y}{\partial y} = \frac{\partial^2 u}{\partial x^2} + \frac{\partial^2 u}{\partial y^2}$.
2. Por Cauchy-Riemann: $\frac{\partial u}{\partial x} = \frac{\partial v}{\partial y}$ y $\frac{\partial u}{\partial y} = -\frac{\partial v}{\partial x}$.
3. Derivando de nuevo:
   $$\frac{\partial^2 u}{\partial x^2} = \frac{\partial^2 v}{\partial x \partial y}, \quad \frac{\partial^2 u}{\partial y^2} = -\frac{\partial^2 v}{\partial y \partial x}$$
4. Por la continuidad de las derivadas parciales (Teorema de Clairaut), $\frac{\partial^2 v}{\partial x \partial y} = \frac{\partial^2 v}{\partial y \partial x}$:
   $$\nabla \cdot \mathbf{V} = \frac{\partial^2 v}{\partial x \partial y} - \frac{\partial^2 v}{\partial y \partial x} = 0 \quad \blacksquare$$

---

### (c) Demostrar que $\nabla \times \mathbf{V} = \mathbf{0}$ (Irrotacionalidad)

**Demostración:**
1. Componente $z$ del rotacional: $(\nabla \times \mathbf{V})_z = \frac{\partial V_y}{\partial x} - \frac{\partial V_x}{\partial y}$.
2. Sustituyendo en términos de $u$:
   $$(\nabla \times \mathbf{V})_z = \frac{\partial^2 u}{\partial x \partial y} - \frac{\partial^2 u}{\partial y \partial x}$$
3. Por el Teorema de Clairaut, $\frac{\partial^2 u}{\partial x \partial y} = \frac{\partial^2 u}{\partial y \partial x}$:
   $$\nabla \times \mathbf{V} = \mathbf{0} \quad \blacksquare$$

```python
# Verificación simbólica con un potencial de prueba f(z) = z^2
x, y = sp.symbols('x y', real=True)
z_expr = x + sp.I*y

# f(z) = z^2 = (x + iy)^2 = (x^2 - y^2) + i(2xy)
f_z = z_expr**2
u = sp.re(f_z.expand())
v = sp.im(f_z.expand())

Vx = sp.diff(u, x)
Vy = sp.diff(u, y)

div_V = sp.diff(Vx, x) + sp.diff(Vy, y)
curl_V = sp.diff(Vy, x) - sp.diff(Vx, y)

print("u(x, y) =", u)
print("v(x, y) =", v)
print("Vx =", Vx, "| Vy =", Vy)
print("Divergencia  ∇ · V =", div_V)
print("Rotacional  (∇ × V)_z =", curl_V)
```

## 3. Teorema Integral de Cauchy: Cálculo de Integral de Contorno

Evaluar la siguiente integral alrededor de un contorno $C$ orientado en sentido antihorario que rodea al punto $z = a$:

$$I = \oint_C \frac{\sin^2 z}{(z - a)^4} \, dz$$

### Solución:
Aplicamos la **Fórmula Integral de Cauchy para Derivadas**:
$$\oint_C \frac{f(z)}{(z - a)^{n+1}} \, dz = \frac{2\pi i}{n!} f^{(n)}(a)$$

Aquí $n = 3$, $n+1 = 4$, y $f(z) = \sin^2 z$:
$$I = \frac{2\pi i}{3!} f'''(a) = \frac{\pi i}{3} f'''(a)$$

Calculamos las derivadas de $f(z) = \sin^2 z$:
1. $f'(z) = 2 \sin z \cos z = \sin(2z)$
2. $f''(z) = 2 \cos(2z)$
3. $f'''(z) = -4 \sin(2z)$

Evaluamos la tercera derivada en $z = a$:
$$f'''(a) = -4 \sin(2a)$$

Sustituyendo en la fórmula integral:
$$I = \frac{\pi i}{3} \left( -4 \sin(2a) \right)$$

$$I = -\frac{4\pi i}{3} \sin(2a)$$
